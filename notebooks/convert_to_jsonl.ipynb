```python

import sys
sys.path.append("..")

from pathlib import Path
import pandas as pd

from src.preprocessing import filter_captions_by_ids

DATA_DIR = Path("../data")
METADATA_PATH = DATA_DIR / "HowTo100M_v1.csv"
CAPTIONS_PATH = DATA_DIR / "caption_clean.json"
OUTPUT_PATH = DATA_DIR / "sports_captions.jsonl"

meta = pd.read_csv(METADATA_PATH, dtype={"video_id": str})
sports = meta[meta["category_1"].str.lower().str.contains("sport", na=False)].copy()
sports_ids = set(sports["video_id"].astype(str))

print(f"Sports videos in metadata: {len(sports_ids):,}")
print(f"Output path: {OUTPUT_PATH}")
```

```python

if OUTPUT_PATH.exists():
    print(f"Already exists: {OUTPUT_PATH}")
else:
    stats = filter_captions_by_ids(CAPTIONS_PATH, OUTPUT_PATH, sports_ids, progress_every=100)
    print(f"Scanned {stats['total']:,} videos, matched {stats['matched']:,} sports captions")
```

```python

# Quick sanity check: JSONL has one line per filtered video.
if OUTPUT_PATH.exists():
    with OUTPUT_PATH.open("rb") as fin:
        line_count = sum(1 for line in fin if line.strip())
    print(f"JSONL lines: {line_count:,}")
    print(f"Expected Sports ids: {len(sports_ids):,}")
```
