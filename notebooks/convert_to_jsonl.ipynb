```python
import os

def fix_nan_in_json(input_path, output_path):
    print("Начинаю очистку файла от NaN... Это может занять время.")
    
    buffer_size = 1024 * 1024 * 10  # 10 MB chunks
    
    with open(input_path, 'r', encoding='utf-8') as fin, \
         open(output_path, 'w', encoding='utf-8') as fout:
        
        leftover = ""
        while True:
            chunk = fin.read(buffer_size)
            if not chunk:
                # Обрабатываем последний остаток
                if leftover:
                    fout.write(leftover.replace('NaN', 'null'))
                break
            
            # Объединяем с остатком от предыдущего куска
            text = leftover + chunk
            
            # Заменяем NaN на null
            # Важно: нужно быть осторожным, чтобы не заменить NaN внутри слова (например, "NaNo-banana")
            # Но в JSON ключи и строки обычно в кавычках, а NaN без кавычек.
            # Простая замена 'NaN' -> 'null' обычно безопасна для JSON, так как NaN всегда отдельный токен.
            
            cleaned_text = text.replace('NaN', 'null')
            
            # Чтобы не разрезать ключи или значения посередине, 
            # можно оставлять небольшой хвост, но для замены NaN это не критично, 
            # так как длина NaN (3) и null (4) близки, но это сдвинет позиции.
            # Для безопасности просто пишем всё, кроме последних 10 символов, которые добираем в следующем цикле?
            # Нет, replace безопасен по содержимому.
            
            fout.write(cleaned_text)
            leftover = ""

    print(f"Очищенный файл сохранен в {output_path}")

# Запуск очистки
input_file = '../data/caption.json'
output_file = '../data/caption_clean.json'

if os.path.exists(input_file):
    fix_nan_in_json(input_file, output_file)
else:
    print("Файл не найден")
```

```python
import json
import ijson

def json_to_jsonl(input_file, output_file):
    with open(input_file, 'rb') as fin, \
        open(output_file, 'w', encoding='utf-8') as fout:
        
        # parser создаёт генератор событий
        objects = ijson.items(fin, 'item') 
        
        count = 0
        for obj in objects:
            # obj - это уже готовый словарь (dict), но в памяти только ОДИН объект
            
            # ТВОЕ УСЛОВИЕ ФИЛЬТРАЦИИ
            if obj.get('category') == 'Sports':
                fout.write(json.dumps(obj, ensure_ascii=False) + '\n')
                count += 1
                
            # Опционально: прогресс-бар
            if count % 1000 == 0:
                print(f"Обработано {count} подходящих записей...")

# Usage
json_to_jsonl('../data/caption_clean.json', '../data/caption.jsonl')
```
